# Principal Component Analysis
This notebook downloads a dataset from Kaggle and trains a model.

```python
!pip install kaggle
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

## 1. Download Dataset


```python
!kaggle datasets download -d c/digit-recognizer --unzip
df = pd.read_csv('train.csv')
df.head()
```

## 2. Preprocess Data


```python
X = df.drop('label', axis=1) / 255.0
```

## 3. Train Model


```python
pca = PCA(n_components=2)X_pca = pca.fit_transform(X)
```

## 4. Evaluate


```python
print('Explained Variance Ratio:', sum(pca.explained_variance_ratio_))plt.scatter(X_pca[:,0], X_pca[:,1], c=df['label'], cmap='tab10', alpha=0.5)plt.colorbar()plt.show()
```
