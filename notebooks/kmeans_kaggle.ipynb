# K-Means Clustering
This notebook downloads a dataset from Kaggle and trains a model.

```python
!pip install kaggle
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## 1. Download Dataset


```python
!kaggle datasets download -d vjchoudhary7/customer-segmentation-tutorial-in-python --unzip
df = pd.read_csv('Mall_Customers.csv')
df.head()
```

## 2. Preprocess Data


```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

## 3. Train Model


```python
model = KMeans(n_clusters=5, random_state=42)model.fit(X)
```

## 4. Evaluate


```python
labels = model.labels_print('Silhouette Score:', silhouette_score(X, labels))plt.scatter(X.iloc[:,0], X.iloc[:,1], c=labels, cmap='viridis')plt.show()
```
