## 1. Download Dataset


```python
!kaggle datasets download -d c/digit-recognizer --unzip
df = pd.read_csv('train.csv')
df.head()
```

## 2. Preprocess Data


```python
X = torch.tensor(df.drop('label', axis=1).values).float().view(-1, 1, 28, 28) / 255.0y = torch.tensor(df['label'].values).long()dataset = TensorDataset(X, y)loader = DataLoader(dataset, batch_size=64, shuffle=True)
```

## 3. Train Model


```python
class CNN(nn.Module):    def __init__(self):        super().__init__()        self.conv = nn.Conv2d(1, 16, 3)        self.fc = nn.Linear(16*26*26, 10)    def forward(self, x):        return self.fc(torch.flatten(torch.relu(self.conv(x)), 1))model = CNN()criterion = nn.CrossEntropyLoss()optimizer = optim.Adam(model.parameters(), lr=0.001)for epoch in range(2):    for data, target in loader:        optimizer.zero_grad()        output = model(data)        loss = criterion(output, target)        loss.backward()        optimizer.step()
```
