# Cross-Validation
Learn robust evaluation techniques to prevent overfitting.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
```

## Load Data

```python
data = load_iris()
X, y = data.data, data.target
model = LogisticRegression(max_iter=1000)
```

## K-Fold Cross Validation
Splits the data into K consecutive folds without shuffling by default.

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(model, X, y, cv=kf)
print('K-Fold Scores:', scores)
print('Mean K-Fold Score:', np.mean(scores))
```

## Stratified K-Fold
Preserves the percentage of samples for each class, which is crucial for imbalanced datasets.

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
stratified_scores = cross_val_score(model, X, y, cv=skf)
print('Stratified K-Fold Scores:', stratified_scores)
print('Mean Stratified Score:', np.mean(stratified_scores))
```
